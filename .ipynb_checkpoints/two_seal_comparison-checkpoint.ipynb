# Two-seal vs one-seal comparison

Tests whether one mud seal of thickness *T* is a conservative upper bound
relative to two mud seals each of thickness *T/2* with the same combined
thickness but separated by a sand gap.

Three runs:
1. **No seal** — baseline
2. **One seal** — thickness `seal_thick`, centred at `seal_depth`
3. **Two seals** — each `seal_thick/2`, properly inserted as two distinct
   MODFLOW layers with a sand interlayer between them

**Fully self-contained** — does not import `pockmark_model.py`.

```python
import contextlib
import io
import warnings
from pathlib import Path
from tempfile import TemporaryDirectory

import flopy
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.filterwarnings('ignore')

# ── Physical constants ────────────────────────────────────────────────────
G            = 9.81
RHO_SEAWATER = 1024.0
PERM_FACTOR  = 1e7 * 365.25 * 24 * 3600   # m² → m/yr
SUBLAYER_THICK = 2.0
N_SUBLAYERS    = 5
MUD_MAX_DEPTH  = N_SUBLAYERS * SUBLAYER_THICK   # 10 m
MUD_FLANK_HALF_WIDTH = 2500.0

print('Imports OK')
```

## Parameters

```python
# ── Shared model parameters ───────────────────────────────────────────────
P = dict(
    overpressure_mpa = 1.55,
    k_mud            = 1e-19 * PERM_FACTOR,
    k_sand           = 1e-11 * PERM_FACTOR,
    ovp_thickness    = 30.0,
    bulkrho_mud      = 1650.0,
    bulkrho_sand     = 2150.0,
    ss               = 1e-4,
    sy               = 0.15,
    Lx               = 105_000.0,
    mud_width        = 35_000.0,
    depth_to_source  = 300.0,
    ncol             = 5000,
)

# ── Seal configuration ────────────────────────────────────────────────────
seal_depth = 150.0   # depth to centre of seal(s) [m]
seal_k     = P['k_mud']
seal_gap   = 20.0    # sand gap between two thin seals [m]

# Figure 1: fixed comparison — no seal vs 1×10 m vs 2×5 m
COMPARE_THICK = 10.0

# Figure 2: one-seal thickness sweep
THICKNESS_SWEEP = [1.0, 10.0, 25.0, 50.0, 100.0]

print(f'Comparison thickness: {COMPARE_THICK} m')
print(f'Thickness sweep: {THICKNESS_SWEEP}')
```

## Simulation functions

```python
def _overpressure_to_head(mpa):
    return mpa / (RHO_SEAWATER * G) * 1e6

def _critical_gradient(bulk_density):
    return (bulk_density - RHO_SEAWATER) / RHO_SEAWATER

def _build_mud_base(ncol, delcol, Lx, mud_width):
    x = (np.arange(ncol) + 0.5) * delcol
    centre = Lx / 2.0
    half   = mud_width / 2.0
    dist   = np.abs(x - centre)
    base   = np.zeros(ncol)
    in_flat  = dist <= (half - MUD_FLANK_HALF_WIDTH)
    in_slope = (dist > (half - MUD_FLANK_HALF_WIDTH)) & (dist <= half)
    base[in_flat] = -MUD_MAX_DEPTH
    if in_slope.any():
        base[in_slope] = (-MUD_MAX_DEPTH / MUD_FLANK_HALF_WIDTH) * (half - dist[in_slope])
    return base

def _harmonic_k(cell_top, cell_bot, mud_base, k_mud, k_sand):
    thick    = cell_top - cell_bot
    mud_t    = np.maximum(0.0, cell_top - np.maximum(mud_base, cell_bot))
    sand_t   = thick - mud_t
    k        = np.full_like(mud_base, k_sand)
    k[sand_t <= 0]                          = k_mud
    mixed = (mud_t > 0) & (sand_t > 0)
    k[mixed] = thick / (mud_t[mixed]/k_mud + sand_t[mixed]/k_sand)
    return k


def _run(
    seals,          # list of (depth_centre, thickness, k) tuples; empty = no seal
    modelname,
    overpressure_mpa, k_mud, k_sand, ovp_thickness,
    bulkrho_mud, bulkrho_sand, ss, sy,
    Lx, mud_width, depth_to_source, ncol,
    perlen_sub=1.0,   nstp_sub=10,
    perlen_init=4.0,  nstp_init=4,
    perlen_early=990.0,  nstp_early=198,
    perlen_late=29000.0, nstp_late=290,
):
    """
    General simulation supporting zero, one, or two seal interbeds.

    `seals` is a list of (centre_depth, thickness, k) tuples, sorted
    shallow-to-deep. Each seal is inserted as its own MODFLOW layer.
    """
    ovp_head   = _overpressure_to_head(overpressure_mpa)
    crit_mud   = _critical_gradient(bulkrho_mud)
    crit_sand  = _critical_gradient(bulkrho_sand)

    delcol   = Lx / ncol
    delrow   = Lx / 3.0
    src_top  = -depth_to_source

    # ── Validate seal geometry ────────────────────────────────────────────
    seals = sorted(seals, key=lambda s: s[0])   # shallow first
    boundaries = [-MUD_MAX_DEPTH]
    for d, t, _ in seals:
        boundaries += [-(d - t/2), -(d + t/2)]
    boundaries += [src_top]
    for i in range(len(boundaries)-1):
        if boundaries[i] <= boundaries[i+1]:
            raise ValueError(
                f'Seal geometry conflict: boundary {boundaries[i]:.1f} <= {boundaries[i+1]:.1f}'
            )

    # ── Build layer bottoms ───────────────────────────────────────────────
    # Structure: N_SUBLAYERS mud/sand sublayers
    #            then for each seal: [sand above, seal, sand below*]
    #            (*sand below is shared with next seal's 'sand above')
    #            ovp source layer
    sand_boundaries = [src_top]   # between-seal and below-seal sand bottoms
    seal_layer_map  = []          # (layer_index, seal_k) for each seal layer

    extra_layers = 1   # bulk sand below all seals (down to src_top) OR just bulk sand
    for d, t, _ in seals:
        extra_layers += 2   # sand-above + seal

    nlay = N_SUBLAYERS + extra_layers + 1   # +1 for ovp source

    botm = np.zeros((nlay, 1, ncol))
    for sl in range(N_SUBLAYERS):
        botm[sl, 0, :] = -(sl + 1) * SUBLAYER_THICK

    # Insert seal layers
    idx = N_SUBLAYERS
    prev_bot = -MUD_MAX_DEPTH   # top of the current sand section
    for seal_i, (d, t, sk) in enumerate(seals):
        seal_top_z = -(d - t/2.0)
        seal_bot_z = -(d + t/2.0)
        botm[idx,     0, :] = seal_top_z   # bottom of sand-above layer
        botm[idx + 1, 0, :] = seal_bot_z   # bottom of seal layer
        seal_layer_map.append((idx + 1, sk))
        idx += 2

    botm[idx,     0, :] = src_top                   # bottom of sand-below / bulk sand
    botm[idx + 1, 0, :] = src_top - ovp_thickness   # bottom of ovp source

    # ── Hydraulic conductivity per layer ──────────────────────────────────
    mud_base = _build_mud_base(ncol, delcol, Lx, mud_width)
    k_layers = []
    for sl in range(N_SUBLAYERS):
        ct = -sl * SUBLAYER_THICK
        cb = -(sl + 1) * SUBLAYER_THICK
        k_layers.append(_harmonic_k(ct, cb, mud_base, k_mud, k_sand))

    seal_k_by_layer = dict(seal_layer_map)
    for i in range(N_SUBLAYERS, nlay - 1):   # everything after sublayers except ovp
        if i in seal_k_by_layer:
            k_layers.append(np.full(ncol, seal_k_by_layer[i]))
        else:
            k_layers.append(np.full(ncol, k_sand))
    k_layers.append(np.full(ncol, k_sand))   # ovp source layer

    assert len(k_layers) == nlay

    # ── MODFLOW-6 setup ───────────────────────────────────────────────────
    tmp  = TemporaryDirectory(delete=False)
    ws   = Path(tmp.name)
    sim  = flopy.mf6.MFSimulation(sim_name=modelname, exe_name='mf6.exe', sim_ws=ws)
    flopy.mf6.ModflowTdis(
        sim, pname='tdis', time_units='YEARS', nper=4,
        perioddata=[
            (perlen_sub,   nstp_sub,   1.0),
            (perlen_init,  nstp_init,  1.0),
            (perlen_early, nstp_early, 1.0),
            (perlen_late,  nstp_late,  1.0),
        ],
    )
    flopy.mf6.ModflowIms(
        sim, pname='ims', complexity='MODERATE',
        linear_acceleration='BICGSTAB',
        outer_maximum=500, inner_maximum=100,
        outer_dvclose=1e-6, inner_dvclose=1e-7, rcloserecord=1e-4,
    )
    gwf = flopy.mf6.ModflowGwf(
        sim, modelname=modelname,
        save_flows=False, newtonoptions='NEWTON UNDER_RELAXATION',
    )
    flopy.mf6.ModflowGwfdis(
        gwf, nlay=nlay, nrow=1, ncol=ncol,
        delr=delrow, delc=delcol, top=0.0, botm=botm,
    )

    # Initial heads: overpressure in source layer, zero elsewhere
    strt = np.zeros((nlay, 1, ncol))
    x    = (np.arange(ncol) + 0.5) * delcol
    strt[nlay - 1, 0, :] = ovp_head   # full-domain overpressure

    flopy.mf6.ModflowGwfic(gwf, pname='ic', strt=strt)
    flopy.mf6.ModflowGwfnpf(gwf, icelltype=1, k=k_layers)
    flopy.mf6.ModflowGwfsto(gwf, ss=ss, sy=sy, transient={i: True for i in range(4)})
    flopy.mf6.ModflowGwfchd(gwf, stress_period_data=[((0, 0, c), 0.0) for c in range(ncol)])
    flopy.mf6.ModflowGwfoc(
        gwf, saverecord=[('HEAD', 'ALL')],
        head_filerecord=[f'{modelname}.hds'],
    )

    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        sim.write_simulation()
        success, buff = sim.run_simulation(silent=False, report=True)
    if not success:
        mf_out = '\n'.join(str(b) for b in buff[-40:]) if buff else buf.getvalue()
        raise RuntimeError(f'MODFLOW did not converge.\n{mf_out}')

    # ── Post-process ──────────────────────────────────────────────────────
    hf        = gwf.output.head()
    times     = np.array(hf.get_times())
    in_mud    = k_layers[0] == k_mud

    rows, grads = [], [np.zeros(ncol)]
    rows.append({'time': 0.0, 'max_gradient_mud': 0.0, 'max_gradient_sand': 0.0})
    for t, ksp in zip(times, hf.get_kstpkper()):
        h    = hf.get_data(kstpkper=ksp)
        grad = (h[1, 0, :] - h[0, 0, :]) / SUBLAYER_THICK
        grads.append(grad)
        rows.append({
            'time':              t,
            'max_gradient_mud':  grad[in_mud].max(),
            'max_gradient_sand': grad[~in_mud].max(),
        })
    hf.file.close()
    try:
        tmp.cleanup()
    except Exception:
        pass

    return pd.DataFrame(rows), crit_mud, crit_sand, np.array(grads), x, in_mud


print('_run() defined')
```

## Run all three cases

```python
import time

def _two_seal_cases(thick):
    half = thick / 2.0
    upper = seal_depth - (half / 2.0 + seal_gap / 2.0)
    lower = seal_depth + (half / 2.0 + seal_gap / 2.0)
    return [(upper, half, seal_k), (lower, half, seal_k)]

# ── Figure 1 runs: no seal / one 10 m / two 5 m ──────────────────────────
fig1_cases = [
    ('No seal',
     [],
     'noseal'),
    (f'One seal ({COMPARE_THICK:.0f} m)',
     [(seal_depth, COMPARE_THICK, seal_k)],
     'oneseal_cmp'),
    (f'Two seals ({COMPARE_THICK/2:.0f}+{COMPARE_THICK/2:.0f} m)',
     _two_seal_cases(COMPARE_THICK),
     'twoseal_cmp'),
]

# ── Figure 2 runs: one seal, varying thickness ────────────────────────────
fig2_cases = [
    (f'One seal ({t:.0f} m)',
     [(seal_depth, t, seal_k)],
     f'oneseal_{int(t):03d}m')
    for t in THICKNESS_SWEEP
]

# Combine, deduplicating by modelname
all_cases = fig1_cases + [c for c in fig2_cases if c[2] not in {x[2] for x in fig1_cases}]

results = {}
for label, seals, name in all_cases:
    if label in results:
        continue
    print(f'Running {label}...')
    t0 = time.time()
    res, crit_mud, crit_sand, gs, x_coords, in_mud = _run(seals=seals, modelname=name, **P)
    results[label] = dict(res=res, gs=gs)
    print(f'  done in {time.time()-t0:.1f} s')

print('\nAll runs complete.')
print(f'Critical gradient — mud: {crit_mud:.4f}  sand: {crit_sand:.4f}')
```

## Figure 1 — No seal vs one 10 m seal vs two 5 m seals

```python
MUD_COLOR  = '#1f77b4'
SAND_COLOR = '#ff7f0e'

f1_styles = {
    fig1_cases[0][0]: dict(color='0.55', ls=':',  lw=1.5),
    fig1_cases[1][0]: dict(color='k',    ls='-',  lw=1.8),
    fig1_cases[2][0]: dict(color='k',    ls='--', lw=1.8),
}

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), constrained_layout=True)
for ax, g_col, crit, zone, color in [
    (axes[0], 'max_gradient_mud',  crit_mud,  'Mud',  MUD_COLOR),
    (axes[1], 'max_gradient_sand', crit_sand, 'Sand', SAND_COLOR),
]:
    for label, _, _ in fig1_cases:
        d = results[label]
        ax.plot(d['res']['time'].values, d['res'][g_col].values,
                label=label, **f1_styles[label])
    ax.axhline(crit, color=color, lw=1.0, ls='-.', alpha=0.8,
               label=f'Critical gradient ({crit:.3f})')
    ax.set_xlabel('Time [yr]')
    ax.set_ylabel('Max exit gradient [-]')
    ax.set_title(f'{zone} zone')
    ax.legend(fontsize=8)
    ax.set_xlim(left=0)

fig.suptitle(
    f'Effect of seal configuration on exit gradient\n',
    fontsize=10,
)
plt.show()
```

## Figure 2 — One seal: thickness sweep (1–100 m)

```python
import matplotlib.cm as cm

cmap   = plt.get_cmap('plasma')
n      = len(THICKNESS_SWEEP)
colors = [cmap(i / (n - 1)) for i in range(n)]

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), constrained_layout=True)
for ax, g_col, crit, zone, crit_color in [
    (axes[0], 'max_gradient_mud',  crit_mud,  'Mud',  MUD_COLOR),
    (axes[1], 'max_gradient_sand', crit_sand, 'Sand', SAND_COLOR),
]:
    # No seal baseline
    ax.plot(results['No seal']['res']['time'].values,
            results['No seal']['res'][g_col].values,
            color='0.55', ls=':', lw=1.5, label='No seal')

    for (label, _, _), color in zip(fig2_cases, colors):
        d = results[label]
        ax.plot(d['res']['time'].values, d['res'][g_col].values,
                color=color, lw=1.6, label=label)

    ax.axhline(crit, color=crit_color, lw=1.0, ls='-.', alpha=0.8,
               label=f'Critical gradient ({crit:.3f})')
    ax.set_xlabel('Time [yr]')
    ax.set_ylabel('Max exit gradient [-]')
    ax.set_title(f'{zone} zone')
    ax.legend(fontsize=8)
    ax.set_xlim(left=0)

fig.suptitle(
    f'Effect of single seal thickness on exit gradient\n'
    f'Seal depth {seal_depth:.0f} m',
    fontsize=10,
)
plt.show()
```
